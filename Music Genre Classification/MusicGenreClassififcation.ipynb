```python
# Performing the 'Label Encoding' to convert labels into integer
from sklearn import preprocessing 
# Creating a preprocessor 'pipeline' to perform such operation
label_encoder = preprocessing.LabelEncoder() 
music_data['label'] = label_encoder.fit_transform(music_data['label'])
```

```python
X = music_data.drop(['label','filename'],axis=1) # Dropping the irrelevant 'filename' column
y = music_data['label']
```

```python
cols = X.columns 
# applies Min-Max scaling to normalize feature values between 0 and 1
minmax = preprocessing.MinMaxScaler() 
np_scaled = minmax.fit_transform(X) 
  
# new data frame with the new scaled data  
X = pd.DataFrame(np_scaled, columns = cols)
feature_columns = X.columns  # From your preprocessed dataset
feature_means = X.mean()     # To fill any missing features
```

```python
# Training the model
from sklearn.model_selection import train_test_split 
  
X_train, X_test, y_train, y_test = train_test_split(X, y,  
                                                    test_size=0.3,  
                                                    random_state=111) 
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
# Testing different algorithms or models to determine the most efficient classifier algorithm
from sklearn.metrics import accuracy_score 
from sklearn.neighbors import KNeighborsClassifier 
from sklearn.tree import DecisionTreeClassifier 
from sklearn.ensemble import RandomForestClassifier 
from sklearn.linear_model import LogisticRegression 

# Initialize classifiers
knn = KNeighborsClassifier(n_neighbors=5)
dt = DecisionTreeClassifier(random_state=0)
rf = RandomForestClassifier(n_estimators=1000, max_depth=10, random_state=0)
lr = LogisticRegression(max_iter=1000)

# Train and evaluate each classifier
for clf in (knn, dt, rf, lr): 
    clf.fit(X_train, y_train) 
    preds = clf.predict(X_test) 
    print(clf.__class__.__name__, accuracy_score(y_test, preds))
```

```python
def extract_features(file_path):
    """
    Extracts a full set of audio features from a given .wav file.
    Returns a DataFrame with all features required for genre prediction.
    """
    y, sr = librosa.load(file_path, mono=True, duration=30)

    # Basic spectral features
    chroma_stft = np.mean(librosa.feature.chroma_stft(y=y, sr=sr))
    rmse = np.mean(librosa.feature.rms(y=y))
    spec_centroid = np.mean(librosa.feature.spectral_centroid(y=y, sr=sr))
    spec_bw = np.mean(librosa.feature.spectral_bandwidth(y=y, sr=sr))
    rolloff = np.mean(librosa.feature.spectral_rolloff(y=y, sr=sr))
    zcr = np.mean(librosa.feature.zero_crossing_rate(y))

    # 20 MFCCs
    mfcc = librosa.feature.mfcc(y=y, sr=sr, n_mfcc=20)
    mfccs = [np.mean(m) for m in mfcc]

    # Assemble all features into a dict
    features = {
        'chroma_stft': chroma_stft,
        'rmse': rmse,
        'spectral_centroid': spec_centroid,
        'spectral_bandwidth': spec_bw,
        'rolloff': rolloff,
        'zero_crossing_rate': zcr,
    }

    for i in range(1, 21):
        features[f'mfcc{i}'] = mfccs[i-1]

    return pd.DataFrame([features])

```

```python
def predict_genre_from_audio(file_path, model, scaler, label_encoder, feature_columns, feature_means):
    """
    Predicts the genre of a .wav file using a trained model.
    Applies feature correction and scaling to match training data.
    """
    try:
        # Extract features from the new audio file
        new_features = extract_features(file_path)

        # Add missing features (if any) using feature_means
        for col in feature_columns:
            if col not in new_features.columns:
                new_features[col] = feature_means[col]

        # Reorder the features to match training order
        new_features = new_features[feature_columns]

        # Scale the features
        new_features_scaled = scaler.transform(new_features)

        # Predict label
        pred_label = model.predict(new_features_scaled)[0]

        # Convert label to genre
        predicted_genre = label_encoder.inverse_transform([pred_label])[0]

        return predicted_genre

    except Exception as e:
        print("Error during prediction:", e)
        return "Prediction failed"
```

```python
file_path = 'genres_original/pop/pop.00008.wav'

predicted_genre = predict_genre_from_audio(
    file_path,
    model=rf,  # or dt, knn, lr
    scaler=minmax,
    label_encoder=label_encoder,
    feature_columns=feature_columns,
    feature_means=feature_means
)

print("Predicted Genre:", predicted_genre)
```
